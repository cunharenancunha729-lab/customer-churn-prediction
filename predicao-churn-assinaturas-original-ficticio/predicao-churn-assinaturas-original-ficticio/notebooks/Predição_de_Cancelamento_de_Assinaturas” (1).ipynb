```python
# Identificando colunas categóricas para One-Hot Encoding (excluindo IDs e colunas já tratadas ou irrelevantes para ML direto)
# Incluindo 'object' e 'category' dtypes para garantir que todas as colunas categóricas sejam consideradas.
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

# Excluir colunas de identificação e outras colunas que não devem ser codificadas diretamente
# Adaptado para os nomes de colunas reais do DataFrame
cols_to_exclude_actual = ['ID', 'NOME', 'ENDEREÇO', 'TELEFONE', 'DATA DO ÚLTIMO PAGAMENTO', 'ÚLTIMO_ACESSO']

categorical_cols = [col for col in categorical_cols if col not in cols_to_exclude_actual]

# Aplicando One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True, dtype=int)

print("Shape do DataFrame após One-Hot Encoding:", df_encoded.shape)
display(df_encoded.head())
```

**Machine Learning**

```python
#importando pandas
import pandas as pd
df_encoded = pd.read_csv('/content/df_encoded.csv')

# Separando X e y

# Definindo a variável alvo 'y' para churn (cancelamento)
# Procuramos por colunas de status geradas pelo One-Hot Encoding que indicam churn.
status_cols_in_encoded_df = [col for col in df_encoded.columns if col.startswith('status_')]

y = None # Inicializa y como None
status_cols_used_for_y = []

# Priorizamos colunas quee indicam um estado de não-ativo/churn
if 'status_Cancelado' in status_cols_in_encoded_df:
    y = df_encoded['status_Cancelado']
    status_cols_used_for_y = ['status_Cancelado']
elif 'status_Inadimplente' in status_cols_in_encoded_df or 'status_Inativo' in status_cols_in_encoded_df:
    # Se houver inadimplência ou inatividade, consideramos churn
    if 'status_Inadimplente' in status_cols_in_encoded_df and 'status_Inativo' in status_cols_in_encoded_df:
        y = ((df_encoded['status_Inadimplente'] == 1) | (df_encoded['status_Inativo'] == 1)).astype(int)
        status_cols_used_for_y = ['status_Inadimplente', 'status_Inativo']
    elif 'status_Inadimplente' in status_cols_in_encoded_df:
        y = df_encoded['status_Inadimplente']
        status_cols_used_for_y = ['status_Inadimplente']
    elif 'status_Inativo' in status_cols_in_encoded_df:
        y = df_encoded['status_Inativo']
        status_cols_used_for_y = ['status_Inativo']
elif 'status_Ativo' in status_cols_in_encoded_df:
    # Se a única coluna de status 'negativa' for 'Ativo', o churn é o oposto (1-Ativo)
    y = 1 - df_encoded['status_Ativo']
    status_cols_used_for_y = ['status_Ativo']
else:
    raise KeyError("Não foi possível identificar uma coluna binária de 'status' em df_encoded para definir a variável alvo 'y'. Verifique a coluna 'status' original e o One-Hot Encoding.")


# Certifique-se de que y foi definido
if y is None:
    raise ValueError("Não foi possível definir a variável alvo 'y'. Por favor, verifique o processamento da coluna 'status'.")


# Lista de colunas a serem removidas das features (X)
# Inclui colunas que são identificadores únicos, ou que são a própria variável alvo/derivadas dela.
cols_to_remove_from_X = []

# Adicionar a coluna de ID do cliente (se existir no df_encoded, o que não deveria ocorrer se foi excluída antes do encoding)
if 'id_cliente' in df_encoded.columns:
    cols_to_remove_from_X.append('id_cliente')

# Adicionar colunas de status que foram usadas para criar 'y' à lista de remoção
cols_to_remove_from_X.extend(status_cols_used_for_y)


# Adicionar colunas relacionadas a 'motivo_cancelamento_' pois são muito próximas do alvo (target leakage)
cols_to_remove_from_X.extend([col for col in df_encoded.columns if col.startswith('motivo_cancelamento_')])


identifier_columns_prefix = [
    'id_cliente_', 'nome_completo_', 'cpf_', 'data_nascimento_', # ID's
    'email_principal_', 'email_alternativo_', 'telefone_celular_', 'telefone_fixo_', # Contato
    'logradouro_', 'complemento_', 'bairro_', 'cidade_', 'estado_uf_', 'cep_', 'pais_', # Endereço
    'data_cadastro_', 'data_ultimo_pagamento_', 'validade_cartao_', 'ultimo_acesso_', 'ip_ultimo_acesso_' # Datas/logs
]

for prefix in identifier_columns_prefix:
    cols_to_remove_from_X.extend([col for col in df_encoded.columns if col.startswith(prefix)])


    cols_to_remove_from_X.append('status')
if 'motivo_cancelamento' in df_encoded.columns:
    cols_to_remove_from_X.append('motivo_cancelamento')

# Remover duplicatas da lista de colunas a serem removidas e garantir que são nomes de colunas válidos
cols_to_remove_from_X = list(set(cols_to_remove_from_X) & set(df_encoded.columns))

# Variáveis de entrada (features)
X = df_encoded.drop(columns=cols_to_remove_from_X, errors='ignore')

# Verificando as dimensões de X e y
print("Shape de X:", X.shape)
print("Shape de y:", y.shape)
```

```python
#Identificando colunas categóricas
categoricas = [
    "Profissão",
    "Plano de Assinatura",
    "Ciclo Mensal",
    "Tipo de Chave Pix ",
    "Bandeira de Cartão",
    "Dispositivo_Principal"
    "indioma_preferido",
   "LEGENDA_PREFERIDA",
   'CONTROLE_PARENTAL_ATIVO',
   "ACEITOU_TERMOS_LGPD",
    'ACEITOU_MARKETING_EMAIL',
    'ACEITOU_MARKETING_SMS',
    'ACEITOU_COOKIES'
     ]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Criando o One Hot Encoder
# transforma as variáveis categóricas em numéricas
preprocessador = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categoricas
        )
    ],
    remainder="passthrough"
)
```

```python
#importando a biblioteca para treinamento do modelo de ml
from sklearn.model_selection import train_test_split

#Dividindo treino e teste
X_train, X_test, y_train, y_test = train_test_split(
X,
y,
test_size=0.20,
random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

# Criando o modelo de Classificação Logística
modelo = LogisticRegression(random_state=42, solver='liblinear')
```

```python
#Treinando o modelo
modelo.fit(X_train, y_train)
```

```python
#Fazendo previsões
y_pred = modelo.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Para modelos de classificação, as previsões devem ser as classes (0 ou 1)
y_pred_logistic = modelo.predict(X_test) # Renomeando para evitar conflito

# Avaliando o modelo de Regressão Logística (agora um classificador)
accuracy_logistic = accuracy_score(y_test, y_pred_logistic)
report_logistic = classification_report(y_test, y_pred_logistic)

print(f"Acurácia (Regressão Logística): {accuracy_logistic:.2f}")
print("\nRelatório de Classificação (Regressão Logística):\n", report_logistic)
```

```python
#Comparando valores reais e previstos
resultado = pd.DataFrame({
    "Cancelamento Real": y_test,
    "Cancelamento Previsto": y_pred
})

resultado.head(10)
```

Comparando Modelos: Árvore de Decisão Na primeira parte  utilizamos a Regressão Linear para prever o cancelamento das assinaturas.
Agora vamos responder à mesma pergunta de negócio utilizando outro algoritmo: Árvore de Decisão. Ao final, compararemos os resultados dos dois modelos utilizando as métricas MAE e R².

```python
#   Criar o modelo de Árvore de Decisão para Classificação
from sklearn.tree import DecisionTreeClassifier
modelo_arvore = DecisionTreeClassifier(random_state=42)
```

```python
#Treinando o modelo
modelo_arvore.fit(
X_train,
y_train
)
```

```python
#Fazendo previsões

y_pred_arvore = modelo_arvore.predict(
X_test
)
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Avaliando o modelo de Árvore de Decisão para Classificação
accuracy_tree = accuracy_score(y_test, y_pred_arvore)
report_tree = classification_report(y_test, y_pred_arvore)

print(f"Acurácia (Árvore de Decisão): {accuracy_tree:.2f}")
print("\nRelatório de Classificação (Árvore de Decisão):\n", report_tree)
```

```python
#Comparando valores reais e previstos
resultado_arvore = pd.DataFrame({
"Cancelamento Real": y_test,
"Cancelamento Previsto": y_pred_arvore
})

resultado_arvore.head(10)
```
